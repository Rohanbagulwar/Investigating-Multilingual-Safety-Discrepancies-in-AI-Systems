```python
import openai
client = openai.OpenAI(
            api_key="API_KEY",
            base_url='https://api.ai.it.ufl.edu'
        )
```

```python
# run with 3 different llms
models=['llama-3.1-70b-instruct', 'mistral-7b-instruct', 'gpt-oss-20b']
import openai
import pandas as pd
```

```python
##change the language
df = pd.read_excel("dataset_tamil-2.xlsx")
reasoning_content_list=[]
responses = []

for prompt in df["tamil_translation"]:
    completion = client.chat.completions.create(
        model="mistral-7b-instruct",
        messages=[
            {"role": "system", "content": "Reply in the same language as the user's prompt. write answer in a short summary."},
            {"role": "user", "content": prompt}
        ]
    )
    reply = completion.choices[0].message.content
    # reasoning_content=completion.choices[0].message.reasoning_content
    responses.append(reply)
    # reasoning_content_list.append(reasoning_content)
    print(reply)

df["response"] = responses

##Chnage the name of file
df.to_excel("output_tamil_mistral.xlsx", index=False)
```
