```python
import pandas as pd
import openai 
import os
from dotenv import load_dotenv
```

```python
NAVIGATOR_API_KEY = os.getenv("NAVIGATOR_API_KEY")
```

```python
client = openai.OpenAI(
    api_key=NAVIGATOR_API_KEY,
    base_url='https://api.ai.it.ufl.edu'
)

df = pd.read_excel("translated_datasets/dataset_Marathi.xlsx")

responses = []

for prompt in df["prompt"]:
    completion = client.chat.completions.create(
        model="llama-3.1-70b-instruct",
        messages=[
            {"role": "system", "content": "Reply in the same language as the user's prompt. write answer in a summary."},
            {"role": "user", "content": prompt}
        ]
    )

    reply = completion.choices[0].message.content
    responses.append(reply)
    print(reply)

df["response"] = responses
df.to_excel("output_marathi_llama.xlsx", index=False)
```
